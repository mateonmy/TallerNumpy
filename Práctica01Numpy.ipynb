1.1 **Listas**

```python
# ── Creación de listas ─────────────────────────────────────────────
numeros = [10, 20, 30, 40, 50]
mixta = [1, 'texto', 3.14, True] # tipos heterogéneos
vacia = []
de_rango = list(range(1, 11)) # [1, 2, 3, ..., 10]
# ── Acceso por índice ──────────────────────────────────────────────
print(numeros[0]) # 10 → primer elemento
print(numeros[-1]) # 50 → último elemento
print(numeros[-2]) # 40 → penúltimo
# ── Slicing [inicio : fin : paso] ───────────────────────────────
# 'fin' es EXCLUIDO — el rango es [inicio, fin)
print(numeros[1:4]) # [20, 30, 40] índices 1,2,3
print(numeros[:3]) # [10, 20, 30] desde el inicio hasta índice 2
print(numeros[2:]) # [30, 40, 50] desde índice 2 hasta el final
print(numeros[::2]) # [10, 30, 50] cada 2 elementos
print(numeros[::-1]) # [50, 40, 30, 20, 10] invertida
# ── Métodos principales ────────────────────────────────────────────
numeros.append(60) # agrega al final → O(1)
numeros.insert(2, 99) # inserta en posición 2 → O(n)
numeros.remove(99) # elimina primera ocurrencia
ultimo = numeros.pop() # extrae y devuelve el último elemento
print(len(numeros)) # longitud de la lista
```

```python
# ── Creación ───────────────────────────────────────────────────────
coordenada = (3.5, -1.2) # tupla de 2 elementos
dimensiones = (28, 28, 3) # alto, ancho, canales (imagen RGB)
unitaria = (42,) # IMPORTANTE: la coma crea la tupla
no_tupla = (42) # esto es sólo el entero 42
# ── Acceso ─────────────────────────────────────────────────────────
print(dimensiones[0]) # 28
print(dimensiones[-1]) # 3
# ── Desempaquetado (unpacking) ─────────────────────────────────────
x, y = coordenada # x = 3.5, y = -1.2
alto, ancho, canales = dimensiones
# ── Las tuplas son inmutables ───────────────────────────────────────
# dimensiones[0] = 64  # ← esto lanzaría TypeError: 'tuple' object does not support item assignment
# ── Uso real: shape de NumPy devuelve tupla ────────────────────────
import numpy as np
A = np.zeros((4, 5))
print(type(A.shape)) # <class 'tuple'>
print(A.shape) # (4, 5)
```

1.3 **Diccionarios**

```python
# ── Creación ───────────────────────────────────────────────────────
estudiante = {
'nombre': 'María',
'edad': 22,
'notas': [8.5, 9.0, 7.5],
'activo': True
}
# ── Acceso ─────────────────────────────────────────────────────────
print(estudiante['nombre']) # 'María' — KeyError si no existe
print(estudiante.get('carrera', 'N/A')) # 'N/A' — valor por defecto seguro
# ── Modificación y adición ─────────────────────────────────────────
estudiante['edad'] = 23 # modifica valor existente
estudiante['semestre'] = 6 # agrega nueva clave
# ── Iteración ──────────────────────────────────────────────────────
for clave, valor in estudiante.items():
    print(f'{clave:12s} → {valor}')
# ── Dict comprehension ─────────────────────────────────────────────
cuadrados = {x: x**2 for x in range(1, 6)}
# {1:1, 2:4, 3:9, 4:16, 5:25}
# ── Verificar existencia de clave ──────────────────────────────────
if 'nombre' in estudiante:
    print('La clave existe')
```

1.4 **Funciones**

```python
# ── Función básica ─────────────────────────────────────────────────
def saludar(nombre):
    return f'Hola, {nombre}!'
# ── Con parámetros por defecto ─────────────────────────────────────
def potencia(base, exponente=2):
    '''Eleva base a exponente. Por defecto calcula el cuadrado.'''
    return base ** exponente
print(potencia(5)) # 25 (usa exponente=2 por defecto)
print(potencia(2, 10)) # 1024
# ── Retorno múltiple (devuelve una tupla) ──────────────────────────
def min_max(lista):
    return min(lista), max(lista) # empaqueta como tupla
minimo, maximo = min_max([3,1,7,2,5]) # desempaqueta
print(minimo, maximo) # 1 7
# ── Lambda — función anónima ────────────────────────────────────────
doble = lambda x: x * 2
print(doble(8)) # 16
# Útil como argumento en funciones de orden superior
nums = [5, 2, 8, 1, 9, 3]
ordenados = sorted(nums, key=lambda x: -x) # orden descendente
# [9, 8, 5, 3, 2, 1]
# ── map y filter ────────────────────────────────────────────────────
cuadrados = list(map(lambda x: x**2, nums))
mayores = list(filter(lambda x: x > 4, nums))
```

**Ejercicio 2: Dado el diccionario datos = {'a':10, 'b':25, 'c':7, 'd':18}** \
a) Crea una lista con las claves cuyo valor sea mayor a 12.\
b) Crea un nuevo diccionario con esas claves y sus valores al cuadrado.


```python
datos = {'a': 10, 'b': 25, 'c': 7, 'd': 18}

# a) Claves cuyo valor es mayor a 12
claves_mayores = [k for k, v in datos.items() if v > 12]
print("Claves con valor > 12:", claves_mayores)

# b) Nuevo diccionario con esas claves y sus valores al cuadrado
dict_cuadrados = {k: datos[k]**2 for k in claves_mayores}
print("Diccionario al cuadrado:", dict_cuadrados)
```

**Ejercicio 3: Define una función 'resumen(lista)' que retorne un diccionario con:** longitud, suma, promedio, mínimo y máximo de la lista.

```python
def resumen(lista):
    return {
        'longitud': len(lista),
        'suma': sum(lista),
        'promedio': sum(lista) / len(lista) if len(lista) > 0 else 0,
        'minimo': min(lista),
        'maximo': max(lista)
    }

# Prueba de la función
valores = [12, 45, 2, 89, 34, 17]
print(resumen(valores))
```

## Módulo 3: Indexing, Slicing, Reshape
Operaciones de sub-indexación N-D, máscaras booleanas, selección condicional (`np.where`) y reestructuración dimensional.

**Ejercicio 1: Dado A = np.arange(1, 26).reshape(5, 5)\
a) Extrae la sub-matriz de las 3 últimas filas y las 3 primeras columnas.\
b) Extrae la segunda columna completa.\
c) Extrae los elementos de los ángulos (esquinas) de la matriz.**

```python
A = np.arange(1, 26).reshape(5, 5)
print("Matriz A:\n", A)

# a) Sub-matriz de 3 últimas filas y 3 primeras columnas
sub_matriz = A[-3:, :3]
print("a) Sub-matriz:\n", sub_matriz)

# b) Segunda columna completa (índice 1)
col_2 = A[:, 1]
print("b) Segunda columna:", col_2)

# c) Elementos de las esquinas (usando indexing con listas)
esquinas = A[[0, 0, -1, -1], [0, -1, 0, -1]]
print("c) Esquinas [sup-izq, sup-der, inf-izq, inf-der]:", esquinas)
```

## Módulo 4: Operaciones Numéricas, Reducciones y Broadcasting
Aplicación de reducciones a lo largo de ejes (`axis`), propagación dimensional de arrays (broadcasting), filtrado analítico y concatenación/división.

**Ejercicio 1: Crea una matriz 5×4 de enteros aleatorios entre 1 y 100 (seed=7).**\
a) Calcula la media, máximo y mínimo de cada columna.\
b) Calcula la suma acumulada de la primera fila.\
c) Normaliza la matriz para que cada columna tenga valores entre 0 y 1.\
Fórmula: (x - min_col) / (max_col - min_col)

```python
np.random.seed(7)
M_rand = np.random.randint(1, 101, size=(5, 4))
print("Matriz:\n", M_rand)

# a) Media, máximo y mínimo por columna (axis=0)
media_col = M_rand.mean(axis=0)
max_col = M_rand.max(axis=0)
min_col = M_rand.min(axis=0)
print("a) Media por columna:", media_col)
print("   Máximo por columna:", max_col)
print("   Mínimo por columna:", min_col)

# b) Suma acumulada de la primera fila
cumsum_fila0 = np.cumsum(M_rand[0, :])
print("b) Suma acumulada fila 0:", cumsum_fila0)

# c) Normalización min-max por columna
M_norm = (M_rand - min_col) / (max_col - min_col)
print("c) Matriz normalizada (0 a 1):\n", M_norm)
```

**Ejercicio 2: Usando broadcasting:**\
a) Crea un vector fila [1,2,3,4] y un vector columna [[10],[20],[30]].\
b) Suma ambos vectores y explica el shape del resultado

```python
# a) Vector fila (1, 4) y vector columna (3, 1)
v_f = np.array([[1, 2, 3, 4]])
v_c = np.array([[10], [20], [30]])

# b) Suma con broadcasting
res_suma = v_f + v_c
print("Resultado suma:\n", res_suma)
print("Shape resultante:", res_suma.shape)
# Explicación: El vector fila (1, 4) se propaga en vertical y el vector
# columna (3, 1) se propaga en horizontal, produciendo una matriz de shape (3, 4).
```

**Ejercicio 4: Crea dos matrices 3×4 y concaténalas vertical y horizontalmente.\
Luego divide el resultado vertical en 3 partes iguales con vsplit.**

```python
M1 = np.ones((3, 4))
M2 = np.full((3, 4), 2)

# Concatenación vertical y horizontal
v_concat = np.vstack((M1, M2))      # Shape (6, 4)
h_concat = np.hstack((M1, M2))      # Shape (3, 8)
print("Vertical concat shape:", v_concat.shape)
print("Horizontal concat shape:", h_concat.shape)

# División vertical en 3 partes iguales (cada bloque tendrá 2 filas x 4 columnas)
partes = np.vsplit(v_concat, 3)
for i, parte in enumerate(partes):
    print(f"Parte {i+1} shape: {parte.shape}")
```
